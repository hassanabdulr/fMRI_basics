# 01 - Concatenate parcellated BOLD TSV files and compute correlations

This notebook demonstrates a basic fMRI workflow for parcellated BOLD timeseries:

1. Collect BOLD TSV files
2. Concatenate runs
3. Compute parcel-by-parcel correlations
4. Save outputs for downstream analysis

```python
from pathlib import Path

import pandas as pd
```

```python
# Update this path to your dataset layout
subject_dir = Path('data/sub-01/func')
pattern = '*_parcellated_bold.tsv'
tsv_files = sorted(subject_dir.glob(pattern))

if not tsv_files:
    raise FileNotFoundError(f'No files found matching {pattern} in {subject_dir}')

print(f'Found {len(tsv_files)} files')
for file in tsv_files:
    print(file)
```

```python
# Read each run and concatenate in time
runs = [pd.read_csv(path, sep='\t') for path in tsv_files]
concatenated = pd.concat(runs, axis=0, ignore_index=True)
concatenated.head()
```

```python
# Parcel-by-parcel correlation matrix
correlation_matrix = concatenated.corr(numeric_only=True)
correlation_matrix
```

```python
# Optional: save outputs
out_dir = Path('outputs/sub-01')
out_dir.mkdir(parents=True, exist_ok=True)

concatenated.to_csv(out_dir / 'concatenated_parcellated_bold.tsv', sep='\t', index=False)
correlation_matrix.to_csv(out_dir / 'parcel_correlation_matrix.tsv', sep='\t')

print(f'Wrote outputs to {out_dir}')
```
