# 02 - Inspect fMRIPrep, FreeSurfer, and XCP-D outputs

This notebook provides a quick orientation to common derivative folders and files generated by:

- fMRIPrep
- FreeSurfer
- XCP-D

Use this to understand what each pipeline produces before running downstream analyses.

```python
from pathlib import Path

derivatives = Path('derivatives')
if not derivatives.exists():
    raise FileNotFoundError('Expected a derivatives/ directory at the project root')

print(f'Derivatives root: {derivatives.resolve()}')
```

```python
# High-level derivative directories
for path in sorted(derivatives.iterdir()):
    if path.is_dir():
        print(path.name)
```

```python
def list_matching_files(base: Path, patterns: list[str], max_results: int = 20):
    matches = []
    for pattern in patterns:
        matches.extend(base.rglob(pattern))
    matches = sorted(set(matches))
    for path in matches[:max_results]:
        print(path)
    if len(matches) > max_results:
        print(f'... and {len(matches) - max_results} more')


# Typical files to inspect in fMRIPrep derivatives
print('\n[fMRIPrep examples]')
list_matching_files(derivatives, ['*desc-preproc_bold.nii.gz', '*desc-confounds_timeseries.tsv'])

# Typical files to inspect in FreeSurfer outputs
print('\n[FreeSurfer examples]')
list_matching_files(derivatives, ['**/freesurfer/**/aparc+aseg.mgz', '**/freesurfer/**/recon-all.log'])

# Typical files to inspect in XCP-D outputs
print('\n[XCP-D examples]')
list_matching_files(derivatives, ['**/xcp_d/**/*_timeseries.tsv', '**/xcp_d/**/*_design.tsv'])
```
